**1)tuple**

**1. Creating a Tuple**

```python
marks=(20,30,40,60,70)
print(marks)
```

```python
fruits=("banana","cherry","mango")
print(fruits)
print(type(fruits))
```

**2. Tuple with Different Data Types**

```python
student=("rahul",21,"surat",True)
print(student)
```

**3. Tuple Indexing**

```python
marks=(20,60,87,98,65,82)
print(marks[0])
print(marks[1])
print(marks[2])
print(marks[-1])
```


**5. Tuple Slicing**

```python
marks=(20,60,87,98,65,82)
print(marks[2:4])
print(marks[5:6])
print(marks[::-1])
```

**6. Tuple Methods**

**count()**

```python

numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))
```

**index()**

```python
numbers=(20,50,87,54,69,85,30)
print(numbers.index(87))
```

**7. Tuple Functions**

**len()**

```python
numbers=(20,50,98,76,43,21)
print(numbers)
print(len(numbers))
```

**max()**

```python
numbers=(20,98,65,74,32)
print(max(numbers))
```

**min()**

```python
numbers=(98,54,78,32,10,12)
print(min(numbers))
```

**sum()**

```python
numbers=(1,2,5,4,3)
print(sum(numbers))
```

**8. Loop Through a Tuple**

```python
fruits=("mango","apple","cherry")
for i in fruits:
    print(i)
```

**9. Check an Element**

```python
fruits=("mango","apple","cherry")
if "banana" in fruits:
    print("found")
else:
    print("not found")
```

**10. Tuple Packing**

```python
subjects="maths","c","java"
print(subjects)
```

**11. Tuple Unpacking**

```python
subjects=("maths",12,"java")

name,rollno,sub=subjects
print(name)
print(rollno)
print(sub)
```

**12. Single-Element Tuple**

```python
number = (10)
print(type(number))
```

**13. Nested Tuple**

```python
marks=(
      ("ram",21),
       ("sita",47),
       ("janki",34),
)
print(marks[0][1])
```

**14. Convert List to Tuple**

```python
data=[20,84,75,96,21]
print(tuple(data))
```

**15. Convert Tuple to List**

```python
data=("rahul",21,"surat")
print(list(data))
```

**8. Count Even and Odd**

```python
numbers = (10,19,20,22)
for i in numbers:

   if i % 2 == 0:
     print(f"{i} is even")
   else: 
    print(f"{i} is odd")
```
