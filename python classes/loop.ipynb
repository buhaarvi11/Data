**1) while Loop**

**1. Print 1 to 10**

```python
i = 1

while i <= 10:
    print(i)
    i = i + 1
```

**2. Print Even Numbers**

```python
i = 1

while i <= 20:
    if i % 2 == 0:
        print(i)
    i = i + 1
```

**3. Sum of Numbers**

```python

n = int(input("Enter a positive integer: "))
total_sum = 0
i = 1

while i <= n:
    total_sum += i 
    i += 1          
print(f"The sum of numbers from 1 to {n} is: {total_sum}")
```

**4. Multiplication Table**

```python
num = int(input("Enter number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```
