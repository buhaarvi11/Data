**Example using a constructor**

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show(self):
        print("name:=",self.name)
        print("age:=",self.age)

s1=student("arvi",21)
s2=student("ravi",25)

s1.show()
s2.show()
```

**self**

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

    def display(self):
        print("employee:",self.name)
        print("salary:",self.salary)

e1=employee("arvi",21000)
e2=employee("kreji",22000)

e1.display()
e2.display()
```

**Instance Attributes**

```python
class Mobile:
      def __init__(self, brand, price):
          self.brand = brand
          self.price = price

mobile1 = Mobile("Samsung", 25000)
mobile2 = Mobile("Apple", 70000)

print(mobile1.brand,mobile1.price)
print(mobile2.brand,mobile2.price)
```
