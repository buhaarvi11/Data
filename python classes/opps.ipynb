**Example using a constructor**

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show(self):
        print("name:=",self.name)
        print("age:=",self.age)

s1=student("arvi",21)
s2=student("ravi",25)

s1.show()
s2.show()
```

**self**

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

    def display(self):
        print("employee:",self.name)
        print("salary:",self.salary)

e1=employee("arvi",21000)
e2=employee("kreji",22000)

e1.display()
e2.display()
```

**Create a Car class with brand, color, and price attributes.**

```python
class car:
    def __init__(self,brand,color,price):

        self.brand=brand
        self.color=color
        self.price=price
        
c1=car("BMW M5","Black",21000000)
c2=car("Mercedes Benz","White",25000000)
c3=car("porche Gt3 rs","Dark Grey",34000000)

print(c1.brand,c1.color,c1.price)
print(c2.brand,c2.color,c2.price)
print(c3.brand,c3.color,c3.price)
```

**Create a Student class with name, roll_number, and marks.**

```python
class student:
    def __init__(self,name,roll_no,marks):

        self.name=name
        self.rollno=roll_no
        self.marks=marks
        
s1=student("arvi","11",95)
s2=student("priya","22",25)
s3=student("raj","33",34)

print(s1.name,s1.rollno,s1.marks)
print(s2.name,s2.rollno,s2.marks)
print(s3.name,s3.rollno,s3.marks)
```

**Create an Employee class with name, department, and salary.**

```python
class employee:
    def __init__(self,name,department,salary):

        self.name=name
        self.department=department
        self.salary=salary
        
e1=employee("arvi","HR",95000)
e2=employee("priya","teacher",2500)
e3=employee("raj","principle",3400)

print(e1.name,e1.department,e1.salary)
print(e2.name,e2.department,e2.salary)
print(e3.name,e3.department,e3.salary)
```

**Create a Product class with name and price. Display the details of three products.**

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_details(self):
        print(f"Product: {self.name} | Price: ${self.price:.2f}")

product1 = Product("Laptop", 999.99)
product2 = Product("Smartphone", 699.50)
product3 = Product("Headphones", 149.99)

product1.display_details()
product2.display_details()
product3.display_details()
```
