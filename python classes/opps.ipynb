**Practical Example 1: Creating a Student Class**

```python
class student:
     def show(self):
        print("i am student")

student1=student()
student2=student()

student1.show()
student2.show()
```

**Student Information**

```python
class student:
    def display(self):
        self.name="arvi"
        self.age="21"

        print("name:=",self.name)
        print("age:=",self.age)

student1=student()
student1.display()
```
