**9. Simple Calculator**

```python
n1=float(input("enter a first number:="))
n2=float(input("enter a second number:="))
operator=input("enter a operator:=")
if operator =="+":
    print("ans:=",n1+n2)
elif operator =="-":
    print("ans:=",n1-n2)
elif operator =="*":
    print("ans:=",n1*n2)
elif operator=="/":
    if n2 == 0:
        print("Cannot divide by zero")
    else:
        print("ans:=",n1/n2)
elif operator=="%":
    print("ans:=",n1%n2)
else:
    print("invalid")
```
