**1.args in Python**

**Example 1: Basic args**

```python
def display(*args):
    print(*args)
display(10,20,30,50)
```


**Example 2: Add any number of values**

```python
def add(*args):
    total = 0
    for number in args:
       total+=number

    return total
print(add(10,20))
print(add(10,50,30))
print(add(50,42,60,85))
```

**Example 3: Find the largest number**

```python
def largest (*args):
    if len(args)==0:
        return
    result=args[0]

    for number in args:
        if number>result:
            result=number
    return result

print(largest(10,50,10,65,98,100))

```
