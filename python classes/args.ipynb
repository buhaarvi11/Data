**1.args in Python**

**Example 1: Basic args**

```python
def display(*args):
    print(*args)
display(10,20,30,50)
```


**Example 2: Add any number of values**

```python
def add(*args):
    total = 0
    for number in args:
       total+=number

    return total
print(add(10,20))
print(add(10,50,30))
print(add(50,42,60,85))
```

**Example 3: Find the largest number**

```python
def largest (*args):
    if len(args)==0:
        return
    result=args[0]

    for number in args:
        if number>result:
            result=number
    return result

print(largest(10,50,10,65,98,100))

```

**kwargs in Python**

**Example 1: Basic kwargs**

```python
def student_info(**kwargs):
 print(kwargs)
student_info(name="Rahul", age=21, course="Python")
```

**Example 2: Display all key-value pairs**

```python
def display(**kwargs):
    for key,Value in kwargs.items():
      print(key,":",Value)
display(name="arvi",age=21,city="surat")
```

**Example 3: Calculate total marks**

```python
def total_marks(**marks):
 total = 0
 for subject, mark in marks.items():
  total += mark
 return total
print(total_marks(math=80, science=90, english=75))

def example(*args, **kwargs):
 print("Args:", args)
 print("Kwargs:", kwargs)
example(10, 20, 30, name="Rahul", age=21)
```

**3. Recursion in Python**

**Example 1: Print numbers from 1 to 5**

```python
def display(n):
 if n == 0:
  return
 display(n - 1)
 print(n)
display(5)
```

**Example 2: Factorial using recursion**

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  return n* factorial(n-1)
print(factorial(6))
```

**Example 3: Sum of numbers from 1 to n**

```python

def sum_numbers(n):
 if n == 0:
  return 0
 return n + sum_numbers(n - 1)
print(sum_numbers(5))
```

**4. Anonymous / Lambda Function**

**Example 1: Add two numbers**

```python
add = lambda a, b: a + b
print(add(10, 20))
```

**Example 2: Square a number**

```python
square = lambda n: n * n
print(square(5))
print(square(8))
```

**Example 3: Even or odd**

```python
check = lambda n: "Even" if n % 2 == 0 else "Odd"
print(check(10))
print(check(7))
```
