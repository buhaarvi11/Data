**Single Inheritance in Python**

```python
class parent:
    def display_name():
        print("name: arvi")


class child(parent):
     def display_course():
        print("course:=c++")


c=child
c.display_course()
c.display_name()
```

```python
class A:
    def display_a():
        print("call A")


class B(A):
    def display_b():
        print("call B")


b=B
b.display_b()
b.display_a()
```

**Multiple Inheritance**

```python
class father:
    def display_f_name():
        print("father name=rajesh")

class mother():
    def display_m_name():
        print("mother name=reema")

class Child(father,mother):
    def display_name():
        print("child name=raj")

c=Child
c.display_f_name()
c.display_m_name()
c.display_name()


```

**Multilevel Inheritance**

```python
class car:
    def display_a():
        print("car = skoda")

class type(car):
    def display_b():
        print("type = 5 seater")

class color(type):
    def display_c():
        print("color = black")

c=color
c.display_a()
c.display_c()
c.display_b()
```

**Hierarchical Inheritance**

```python
class A:
    def display_a():
        print("call A")


class B(A):
    def display_b():
        print("call B")

class C(A):
    def display_c():
        print("call C")

class D(A):
    def display_d():
        print("call D")


b=B
b.display_a()
b.display_b()


c=C
c.display_a()
c.display_c()

d=D
d.display_a()
d.display_d()
```

**Hybrid Inheritance**

```python
class A:
    def display_a():
        print("call A")

class B(A):
    def display_b():
        print("call B")

class C:
     def display_c():
        print("call C")

class D(B,C):
     def display_d():
        print("call D")




d=D
d.display_a()
d.display_b()
d.display_c()
d.display_d()
```
