**Single Inheritance in Python**

```python
class parent:
    def display_name():
        print("name: arvi")


class child(parent):
     def display_course():
        print("course:=c++")


c=child
c.display_course()
c.display_name()
```

```python
class A:
    def display_a():
        print("call A")


class B(A):
    def display_b():
        print("call B")


b=B
b.display_b()
b.display_a()
```

**Multiple Inheritance**

```python
class father:
    def display_f_name():
        print("father name=rajesh")

class mother():
    def display_m_name():
        print("mother name=reema")

class Child(father,mother):
    def display_name():
        print("child name=raj")

c=Child
c.display_f_name()
c.display_m_name()
c.display_name()


```

**Multilevel Inheritance**

```python
class A:
    def display_a():
        print("called A")

class B(A):
    def display_b():
        print("called B")

class C(B):
    def display_c():
        print("called C")

c=C
c.display_a()
c.display_c()
c.display_b()
```
