**match-case**

**6. Day Program**

```python
day= int(input("Enter a number from 1 to 7: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid input")
```

**7. Calculator**

```python
n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

match operator:
    case "+":
        print("Result =", n1 + n2)

    case "-":
        print("Result =", n1 - n2)

    case "*":
        print("Result =", n1 * n2)

    case "/":
        if n2 != 0:
            print("Result =", n1 / n2)
        else:
            print("Cannot divide by zero")

    case _:
        print("Invalid operator")
```

**8. Menu Program**

```python
print("1. Add")
print("2. View")
print("3. Update")
print("4. Delete")
print("5. Exit")

choice = int(input("Enter your choice: "))

match choice:
    case 1:
        print("Add")

    case 2:
        print("View")

    case 3:
        print("Update")

    case 4:
        print("Delete")

    case 5:
        print("Exit")

    case _:
        print("Invalid choice")
```

**10. Traffic Light**

```python
color = input("Enter color: ")

match color:
    case "red":
        print("Stop")

    case "yellow":
        print("Wait")

    case "green":
        print("Go")

    case _:
        print("Invalid color")
```
