**Complete Dice Rolling Game**

```python
import random

while True:
    choice = input("\nDo you want to roll the dice? (yes/no): ")

    if choice.lower() == "yes":
        dice = random.randint(1, 6)
        print("You rolled:", dice)

    elif choice.lower() == "no":
        print("Thank you")
        break

    else:
        print("Invalid choice")
```
