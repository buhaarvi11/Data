**12. Function With List**

```python
def find_sum(numbers):
  total = 0
  for number in numbers:
   total += number
  return total
numbers = [10, 20, 30, 40]

print(find_sum(numbers))
```

**13. Function With Dictionary**

```python
def display_student(student):
 for key, value in student.items():
   print(key, ":", value)
student = {
"name": "Rahul",
"age": 21,
"course": "Python"
}
display_student(student)
```

```python
def factorial(n):
 result = 1
 for i in range(1, n + 1):
  result = result * i
 return result
print(factorial(5))
```

**20. Function for Largest Number**

```python
def find_largest(numbers):
 largest = numbers[0]
 for number in numbers:
   if number > largest:
     largest = number
 return largest

numbers = [10, 45, 23, 78, 12]
print(find_largest(numbers))
```

**13. Function to count even and odd numbers in a list**

```python
def even_odd(numbers):
    even=0
    odd=0
    for number in numbers:
        if number % 2==0:
            even+=1
        else:
            odd+=1
    return even,odd
numbers=[13,60,42,65,98,72]

even,odd=even_odd(numbers)

print("even:",even)
print("odd:",odd)
```

**14. Function to find the largest element in a list**

```python
def find_largest(numbers):
  largest = numbers[0]

  for number in numbers:
       if number > largest:
        largest = number
  return largest

numbers = [10, 45, 23, 78, 12]
print(find_largest(numbers))
```
