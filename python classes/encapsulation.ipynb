**Encapsulation**

```python
class employee:
    def __init__(self,name,salary):
        self.name=name #public 
        self.__salary=salary   #private     

emp=employee("arvi",20000)
print(emp.name)
print(emp.__salary)
```

**public members**

```python
class employee:
    def __init__(self,name):
        self.name=name

    def display_name(self):
        print(self.name)

emp=employee("arvi")

print(emp.name)

```

**Protected members**

```python
class employee:
    def __init__(self,name,age):
        self.name=name#public
        self._age=age#protected

class subemployee(employee):
    def show_age(self):
        print("age:",self._age)

emp=subemployee("arvi",21)
print(emp.name)
emp.show_age()
```

**Private members**

```python
class employee:
    def __init__(self,name,age):
        self.name=name
        self.__age=age

    def show_age(self):
        print("age:",self.__age)

emp=employee("arvi",21)
print(emp.name)
emp.show_age()
```
