**18. Find Second Largest**

```python
numbers = [10, 50, 20, 40, 30]
largest = numbers[0]
second_largest = numbers[0]
for number in numbers:
   if number > largest:
       second_largest = largest
       largest = number
   elif number > second_largest and number != largest:
            second_largest = number

print("Largest:", largest)
print("Second largest:", second_largest)
```

**27. Move Zeros to the End**

```python
numbers = [0, 10, 0, 20, 30, 0, 40]
result = []
zero_count = 0

for number in numbers:
  if number == 0:
    zero_count += 1
  else:
      result.append(number)
for i in range(zero_count):
       result.append(0)

print(result)
```

**37. Check if List is Sorted**

```python
numbers = [10, 20, 30, 40, 50]
sorted_list = True
for i in range(len(numbers) - 1):

     if numbers[i] > numbers[i + 1]:
          sorted_list = False
     break
if sorted_list:
   print("List is sorted")
else:
   print("List is not sorted")
```

**38. Find Missing Number**

```python
numbers = [1, 2, 3, 4, 5, 7, 8, 9, 10]
expected = 0
actual = 0
for i in range(1, 11):
   expected += i
for number in numbers:
    actual += number
missing = expected - actual
print("Missing number =", missing)
```

**39. Check Palindrome List**

```python
numbers = [1, 2, 3, 2, 1]
palindrome = True
for i in range(len(numbers) // 2):
   if numbers[i] != numbers[len(numbers) - 1 - i]:
     palindrome = False
   break
if palindrome:
   print("Palindrome")
else:
   print("Not Palindrome")
```

**40. Merge Two Lists Without Duplicates**

```python
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]
result = []

for number in list1:
   if number not in result:
     result.append(number)
for number in list2:
  if number not in result:
    result.append(number)
print(result)
```
