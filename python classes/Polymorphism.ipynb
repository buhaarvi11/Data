**3. Runtime Polymorphism (Method Overriding)**

```python
class animal:
    def speak(self):
        return "Generic animal sound"

class dog:
    def speak(self):
        return "Bark!"
    
class cat:
    def speak(self):
        return "Meow!"


animals = [dog(), cat(), animal()]
for animal in animals:
     print(animal.speak())
```
