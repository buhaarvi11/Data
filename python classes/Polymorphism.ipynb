**3. Runtime Polymorphism (Method Overriding)**

```python
class animal:
    def speak(self):
        return "Generic animal sound"

class dog:
    def speak(self):
        return "Bark!"
    
class cat:
    def speak(self):
        return "Meow!"


animals = [dog(), cat(), animal()]
for animal in animals:
     print(animal.speak())
```

**Method Overloading (Compile-Time)**

```python
class Calculator:
    def add(self, a, b, c=0):
      return a + b + c

calc = Calculator()
print(calc.add(2, 3))
print(calc.add(2, 3, 4))
```
