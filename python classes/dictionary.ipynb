**12. Find the student with the highest marks**

```python
students = {
"Rahul": 55,
"Amit": 82,
"Priya": 91,
"Neha": 75,
"Ravi": 78
}
highest_marks = 0
highest_student = ""

for name , marks in students.items():
    if marks > highest_marks:
        highest_marks=marks
        highest_student=name

print("student:",highest_student)
print("marks:",highest_marks)

```

**13. Find the student with the lowest marks**

```python
students = {
"Rahul": 55,
"Amit": 82,
"Priya": 91,
"Neha": 75,
"Ravi": 78
}
lowest_marks = 101
lowest_student = ""

for name , marks in students.items():
    if marks < lowest_marks:
        lowest_marks=marks
        lowest_student=name

print("student:",lowest_student)
print("marks:",lowest_marks)
```

**14. Count how many students passed and failed**

```python
students = {
"Rahul": 85,
"Amit": 35,
"Priya": 91,
"Neha": 28,
"Ravi": 78
}
passed = 0

failed = 0
for name, marks in students.items():
 if marks >= 40:
  passed += 1
else:
  failed += 1
print("Passed:", passed)
print("Failed:", failed)
```
