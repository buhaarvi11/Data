**Python Modules: datetime, math, and random**

**1. datetime Module**

```python
import datetime
today=datetime.date.today()
print(today)
```

**2. Get current date and time**

```python
now=datetime.datetime.now()
print(now)
```

**3. Get current year, month and day**

```python
today=datetime.date.today()
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

**4. Create a specific date**

```python
my_date=datetime.date(2026,12,5)
print(my_date)
```

**5. Create a specific time**

```python
my_time=datetime.time(3,12,53)
print(my_time)
```

**6. Format date using strftime()**

```python
now = datetime.datetime.now()
print(now.strftime("%d-%m-%Y"))
print(now.strftime("%d/%m/%Y"))
print(now.strftime("%B"))
print(now.strftime("%A"))
print(now.strftime("%H:%M:%S"))
```

**7. Convert string into date using strptime()**

```python
date_string="26-11-2026 14:20:50"
result=datetime.datetime.strptime(date_string,"%d-%m-%Y %H:%M:%S")
print(result)
```

**8. Date difference using timedelta**

```python
today=datetime.date.today()

future_date=today+datetime.timedelta(days=10)
print("today:",today)
print("After 10 days:", future_date)
```

**math Module**

**1. Square root – sqrt()**

```python
import math
print(math.sqrt(25))
print(math.sqrt(100))
```

**2. Power – pow()**

```python
print(math.pow(3,3))
```

**3. Ceiling – ceil()**

```python
print(math.ceil(3.4))
print(math.ceil(10.5))
```

**4. Floor – floor()**

```python
print(math.floor(4.7))
print(math.floor(10.7))
```

**5. Absolute value – fabs()**

```python
print(math.fabs(-19))
```

**6. Factorial – factorial()**

```python
print(math.factorial(5))
```

**random Module**

**1. Generate random integer – randint()**

```python
import random
num=random.randint(1,10)
print(num)
```

**2. Generate random float – random()**

```python
num=random.random()
print(num)
```

**3. Generate random number using uniform()**

```python
num=random.uniform(10,50)
print(num)
```

**4. Select random element – choice()**

```python
list=["arvi","khushi","kreji"]

result = random.choice(list)

print(result)
```

**5. Shuffle list – shuffle()**

```python
numbers=[1,2,3,4,5]
random.shuffle(numbers)
print(numbers)
```

**7. Random number using randrange()**

```python

print(random.randrange(1, 20, 2))
```
