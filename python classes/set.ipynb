**Python Set**

**1. Creating a Set**

```python
no={10,50,84,75,96,32}
print(no)
```

**2. Duplicate Values**

```python
no={10,50,84,10,26,10}
print(no)
```

**3. Empty Set**

```python
marks={}
print(marks)
```

**Set Methods 4. add()**

```python
marks={"banana",21,"market"}
marks.add(60)
print(marks)
```

**5. update()**

```python
no={50,98,75,65,32}
no.update([50,10,20])
print(no)
```

**6. remove()**

```python
marks={10,85,42,62}
marks.remove(85)
print(marks)
```

**7. discard()**

```python
marks={10,85,42,62}
marks.discard(100)
print(marks)
```

**8. pop()**

```python
marks={10,85,42,62}
marks.pop()
print(marks)
```

**9. clear()**

```python
marks={10,85,42,62}
marks.clear()
print(marks)
```

**Set Operations 10. Union |**

```python
a={1,2,3}
b={5,6,7}
#result=a|b
result = a.union(b)
print(result)
```

**11. Intersection &**

```python
a={1,2,3}
b={2,6,7}
result=a&b
#result = a.intersection(b)
print(result)
```

**12. Difference -**

```python
a={1,2,3}
b={2,3,7}
#result=a-b
result = a.difference(b)
print(result)
```

**13. Symmetric Difference ^**

```python
a={1,2,3}
b={2,3,7}
#result=a^b
result = a.symmetric_difference(b)
print(result)
```

**Set Membership**

```python
numbers = {10, 20, 30, 40}
print(20 in numbers)
print(50 in numbers)
```

**Loop Through Set**

```python
numbers = {10, 20, 30, 40}
for i in numbers:
  print(i)
```

**Set Comparison issubset()**

```python
a={2,3}
b={1,2,3,4}

print(a.issubset(b))
```

**issuperset()**

```python
a={1,2,3,4}
b={2,3}

print(a.issuperset(b))
```

**isdisjoint()**

```python
a={1,2,3}
b={2,3,7}

print(a.isdisjoint(b))
```

**extra example**

```python
my_list = [10, 20, 10, 30]
my_tuple = (10, 20, 10, 30)
my_set = {10, 20, 10, 30}
print(my_list)
print(my_tuple)
print(my_set)
```

**Convert List to Set**

```python
marks=[10,85,74,95,65]
print(set(marks))
```
